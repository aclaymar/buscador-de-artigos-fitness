## 🔌 Cliente OpenAlex

```
"""
Cliente da API OpenAlex (espelha src/openalex_client.py).

Coleta paginada via cursor, com retry exponencial e polite pool via mailto.
Documentação: https://docs.openalex.org/
"""
import time
import logging
from typing import Optional
from dataclasses import dataclass, field

import requests
import backoff
from tqdm.auto import tqdm

logger = logging.getLogger(__name__)

OPENALEX_BASE_URL = "https://api.openalex.org/works"
OPENALEX_PER_PAGE_MAX = 200
DEFAULT_PER_PAGE = 50
DEFAULT_TIMEOUT = 30

DEFAULT_SELECT_FIELDS = [
    "id", "doi", "title", "publication_year", "publication_date",
    "type", "language", "open_access",
    "authorships", "primary_location",
    "cited_by_count", "concepts", "topics",
    "abstract_inverted_index", "is_retracted",
]


@dataclass
class OpenAlexArticle:
    """Registro simplificado de um trabalho retornado pela OpenAlex."""
    id: str
    doi: Optional[str]
    titulo: str
    ano: Optional[int]
    data_publicacao: Optional[str]
    idioma: Optional[str]
    tipo: Optional[str]
    acesso_aberto: bool
    retratado: bool
    primeiro_autor: Optional[str]
    autores: list = field(default_factory=list)
    n_autores: int = 0
    instituicoes: list = field(default_factory=list)
    paises: list = field(default_factory=list)
    pais_principal: Optional[str] = None
    periodico: Optional[str] = None
    citacoes: int = 0
    conceitos: list = field(default_factory=list)
    topicos: list = field(default_factory=list)
    abstract: str = ""
    queries_origem: list = field(default_factory=list)
    areas_origem: list = field(default_factory=list)

    def to_dict(self) -> dict:
        return {
            "id": self.id,
            "doi": self.doi,
            "titulo": self.titulo,
            "ano": self.ano,
            "data_publicacao": self.data_publicacao,
            "idioma": self.idioma,
            "tipo": self.tipo,
            "acesso_aberto": self.acesso_aberto,
            "retratado": self.retratado,
            "primeiro_autor": self.primeiro_autor,
            "autores": "; ".join(self.autores),
            "n_autores": self.n_autores,
            "instituicoes": "; ".join(self.instituicoes),
            "paises": "; ".join(self.paises),
            "pais_principal": self.pais_principal,
            "periodico": self.periodico,
            "citacoes": self.citacoes,
            "conceitos": "; ".join(self.conceitos),
            "topicos": "; ".join(self.topicos),
            "abstract": self.abstract,
            "queries_origem": "; ".join(self.queries_origem),
            "areas_origem": "; ".join(self.areas_origem),
        }


class OpenAlexClient:
    """Cliente para a API REST de works da OpenAlex."""

    def __init__(self, mailto: str = "", timeout: int = DEFAULT_TIMEOUT):
        self.mailto = mailto
        self.timeout = timeout
        self.session = requests.Session()
        self.session.headers.update({
            "User-Agent": f"Fitness-OpenAlex (mailto:{mailto})" if mailto
                          else "Fitness-OpenAlex",
        })

    @staticmethod
    def reconstruir_abstract(inverted_index: Optional[dict]) -> str:
        """Reconstrói o texto a partir do inverted index {palavra: [posições]}."""
        if not inverted_index:
            return ""
        try:
            posicoes = []
            for palavra, lista_pos in inverted_index.items():
                for p in lista_pos:
                    posicoes.append((p, palavra))
            posicoes.sort(key=lambda x: x[0])
            return " ".join(palavra for _, palavra in posicoes)
        except Exception as e:
            logger.warning(f"Falha ao reconstruir abstract: {e}")
            return ""

    def _parse_artigo(self, registro: dict) -> OpenAlexArticle:
        autorias = registro.get("authorships", []) or []
        autores = [a["author"]["display_name"] for a in autorias
                   if a.get("author") and a["author"].get("display_name")]
        primeiro_autor = autores[0] if autores else None

        instituicoes, paises, pais_principal = [], [], None
        if autorias:
            for inst in autorias[0].get("institutions", []) or []:
                if inst.get("display_name"):
                    instituicoes.append(inst["display_name"])
                if inst.get("country_code"):
                    paises.append(inst["country_code"])
            pais_principal = paises[0] if paises else None

        primary_location = registro.get("primary_location") or {}
        source = primary_location.get("source") or {}
        periodico = source.get("display_name")

        oa = registro.get("open_access") or {}
        acesso_aberto = bool(oa.get("is_oa", False))

        conceitos = [c["display_name"]
                     for c in (registro.get("concepts", []) or [])[:5]
                     if c.get("display_name")]
        topicos = [t["display_name"]
                   for t in (registro.get("topics", []) or [])[:5]
                   if t.get("display_name")]

        return OpenAlexArticle(
            id=registro.get("id", ""),
            doi=registro.get("doi"),
            titulo=registro.get("title", "") or "",
            ano=registro.get("publication_year"),
            data_publicacao=registro.get("publication_date"),
            idioma=registro.get("language"),
            tipo=registro.get("type"),
            acesso_aberto=acesso_aberto,
            retratado=bool(registro.get("is_retracted", False)),
            primeiro_autor=primeiro_autor,
            autores=autores,
            n_autores=len(autores),
            instituicoes=instituicoes,
            paises=paises,
            pais_principal=pais_principal,
            periodico=periodico,
            citacoes=registro.get("cited_by_count", 0) or 0,
            conceitos=conceitos,
            topicos=topicos,
            abstract=self.reconstruir_abstract(
                registro.get("abstract_inverted_index")
            ),
        )

    @backoff.on_exception(
        backoff.expo,
        (requests.exceptions.RequestException,),
        max_tries=5,
        max_time=120,
    )
    def _fetch_pagina(self, params: dict) -> dict:
        response = self.session.get(
            OPENALEX_BASE_URL, params=params, timeout=self.timeout,
        )
        response.raise_for_status()
        return response.json()

    def search(
        self,
        query: str,
        ano_inicio: int = 2015,
        ano_fim: int = 2025,
        max_results: int = 200,
        idiomas: Optional[list] = None,
        tipos: Optional[list] = None,
        per_page: int = DEFAULT_PER_PAGE,
        progress: bool = True,
    ) -> list:
        """
        Busca paginada por cursor. Retorna lista de OpenAlexArticle.

        idiomas=None significa sem filtro de idioma (qualquer língua).
        """
        if tipos is None:
            tipos = ["article", "review"]

        filters = [
            f"from_publication_date:{ano_inicio}-01-01",
            f"to_publication_date:{ano_fim}-12-31",
            f"type:{'|'.join(tipos)}",
            "has_abstract:true",
        ]
        if idiomas:
            filters.append(f"language:{'|'.join(idiomas)}")
        params = {
            "search": query,
            "filter": ",".join(filters),
            "per-page": min(per_page, OPENALEX_PER_PAGE_MAX),
            "select": ",".join(DEFAULT_SELECT_FIELDS),
            "cursor": "*",
        }
        if self.mailto:
            params["mailto"] = self.mailto

        artigos = []
        pbar = tqdm(total=max_results, desc="OpenAlex", disable=not progress)

        while len(artigos) < max_results:
            data = self._fetch_pagina(params)
            results = data.get("results", []) or []
            if not results:
                break
            for registro in results:
                if len(artigos) >= max_results:
                    break
                try:
                    artigos.append(self._parse_artigo(registro))
                    pbar.update(1)
                except Exception as e:
                    logger.warning(f"Falha ao processar registro: {e}")

            next_cursor = data.get("meta", {}).get("next_cursor")
            if not next_cursor:
                break
            params["cursor"] = next_cursor
            time.sleep(0.1)

        pbar.close()
        logger.info(f"Coletados {len(artigos)} trabalhos para: '{query[:50]}'")
        return artigos
```

## 🧠 Processamento

```
"""
Processamento dos trabalhos coletados.

Etapas:
    1. Deduplicação por OpenAlex ID (guarda quais queries e áreas encontraram cada trabalho)
    2. Filtragem: abstract mínimo, exclusão de retratados (e foco técnico, se ativado)
    3. Classificação por área temática (palavras-chave, com limites de palavra)
    4. Classificação do desenho do estudo e nível de evidência
"""
import re
import logging
from collections import Counter

import pandas as pd


logger = logging.getLogger(__name__)


# =============================================================
# DEDUPLICAÇÃO
# =============================================================
def deduplicar(artigos_por_query: dict, areas_por_query: dict) -> list:
    """
    Remove duplicatas mantendo as queries e as áreas que recuperaram cada trabalho.

    Args:
        artigos_por_query: {codigo_query: [OpenAlexArticle, ...]}
        areas_por_query:   {codigo_query: "nome da área"}
    """
    unicos = {}
    for codigo, lista in artigos_por_query.items():
        area = areas_por_query.get(codigo, "")
        for artigo in lista:
            if artigo.id not in unicos:
                artigo.queries_origem = [codigo]
                artigo.areas_origem = [area]
                unicos[artigo.id] = artigo
            else:
                existente = unicos[artigo.id]
                if codigo not in existente.queries_origem:
                    existente.queries_origem.append(codigo)
                if area not in existente.areas_origem:
                    existente.areas_origem.append(area)

    total = sum(len(v) for v in artigos_por_query.values())
    logger.info(f"Deduplicação: {total} → {len(unicos)} trabalhos únicos")
    return list(unicos.values())


# =============================================================
# FILTRAGEM
# =============================================================
_CJK = re.compile(r"[぀-ヿ㐀-鿿가-힯]")


def _contem_termo(texto_lower: str, termo: str) -> bool:
    """
    Busca o termo no texto.

    Para termos em alfabeto latino/cirílico usa limite de palavra (evita
    falsos positivos, ex.: 'run' em 'brunch'). Para chinês, japonês e coreano,
    que não separam palavras por espaço, usa busca por substring.
    """
    if _CJK.search(termo):
        return termo in texto_lower
    return re.search(rf"\b{re.escape(termo)}\b", texto_lower) is not None


def tem_foco_tecnico(texto: str) -> bool:
    t = texto.lower()
    return any(_contem_termo(t, termo) for termo in TERMOS_TECNICOS)


def filtrar_artigos(artigos: list) -> tuple:
    """
    Critérios de inclusão/exclusão:
      - abstract com pelo menos MIN_CARACTERES_ABSTRACT caracteres
      - não retratado
      - (opcional) sem foco puramente técnico

    Returns:
        (incluidos, contagem_de_exclusoes)
    """
    incluidos = []
    excl = Counter()

    for artigo in artigos:
        if not artigo.abstract or len(artigo.abstract) < MIN_CARACTERES_ABSTRACT:
            excl["sem abstract válido"] += 1
            continue
        if artigo.retratado:
            excl["retratado"] += 1
            continue
        if FILTRO_TECNICO and tem_foco_tecnico(f"{artigo.titulo} {artigo.abstract}"):
            excl["foco técnico"] += 1
            continue
        incluidos.append(artigo)

    logger.info(f"Filtragem: {len(artigos)} → {len(incluidos)} incluídos ({dict(excl)})")
    return incluidos, dict(excl)


# =============================================================
# CLASSIFICAÇÃO POR ÁREA
# =============================================================
def classificar_area(texto: str) -> tuple:
    """
    Retorna (área principal, áreas secundárias, contagens).
    A área principal é a de maior número de termos encontrados.
    """
    t = texto.lower()
    contagens = {}
    for area, termos in DICIONARIO_AREAS.items():
        n = sum(1 for termo in termos if _contem_termo(t, termo))
        if n > 0:
            contagens[area] = n

    if not contagens:
        return "Outros", [], {}

    ordenadas = sorted(contagens.items(), key=lambda x: x[1], reverse=True)
    principal = ordenadas[0][0]
    secundarias = [a for a, _ in ordenadas[1:]]
    return principal, secundarias, contagens


# =============================================================
# CLASSIFICAÇÃO DO DESENHO DO ESTUDO
# =============================================================
def classificar_desenho(texto: str) -> tuple:
    """Retorna (desenho, nivel_evidencia) pelo primeiro padrão que bate."""
    t = texto.lower()
    for nome, nivel, padroes in DESENHOS_ESTUDO:
        if any(_contem_termo(t, p) for p in padroes):
            return nome, nivel
    return DESENHO_NAO_IDENTIFICADO


# =============================================================
# DATAFRAME FINAL
# =============================================================
def aplicar_classificacoes(artigos: list) -> pd.DataFrame:
    registros = []
    for artigo in artigos:
        texto = f"{artigo.titulo} {artigo.abstract}"
        area, secundarias, _ = classificar_area(texto)
        desenho, nivel = classificar_desenho(texto)

        registro = artigo.to_dict()
        registro["area_principal"] = area
        registro["areas_secundarias"] = "; ".join(secundarias)
        registro["desenho_estudo"] = desenho
        registro["nivel_evidencia"] = nivel
        registros.append(registro)

    df = pd.DataFrame(registros)
    if not df.empty:
        logger.info("Distribuição por área:\n" + df["area_principal"].value_counts().to_string())
    return df


# =============================================================
# PIPELINE COMPLETO
# =============================================================
def processar_pipeline(artigos_por_query: dict, areas_por_query: dict) -> tuple:
    """
    Executa deduplicar → filtrar → classificar.

    Returns:
        (DataFrame final, dict com contagem de exclusões)
    """
    print("\n" + "=" * 60)
    print("PIPELINE DE PROCESSAMENTO")
    print("=" * 60)

    print("\n[1/3] Deduplicação...")
    unicos = deduplicar(artigos_por_query, areas_por_query)

    print("\n[2/3] Filtragem...")
    incluidos, exclusoes = filtrar_artigos(unicos)

    print("\n[3/3] Classificação por área e desenho do estudo...")
    df = aplicar_classificacoes(incluidos)

    print(f"\n✓ Pipeline concluído: {len(df)} trabalhos finais")
    return df, exclusoes
```
